## Question 8 — Running total with a condition (6 points)

Write `sum_multiples(limit)`.

- Visit every integer from `1` through `limit`, inclusive.
- Add only numbers divisible by `3`.
- Use a running total initialized before the loop.
- Return the final total.

Example:

```python
sum_multiples(10) == 18
```


````python
# Question 8 code
def sum_multiples(limit):
    total = 0
    for i in range (1, limit + 1):
        if i % 3 == 0:
            total+= i
    return total
````

````python
# Question 8 public tests
assert sum_multiples(10) == 18
assert sum_multiples(0) == 0
assert sum_multiples(3) == 3
print("Question 8 tests passed.")
````

## Question 9 — Filtering a list with `enumerate()` (6 points)

Write `passing_positions(values, cutoff)`.

- Use `enumerate(values, start=1)`.
- Keep values greater than or equal to the cutoff.
- Append `(position, value)` tuples to a new list.
- Return the new list.

For:

```python
values = [65, 90, 45, 82, 73]
cutoff = 70
```

predict the returned list before coding: ______________________________


````python
# Question 9 code
def passing_positions(values, cutoff):
    result = []
    for position,value in enumerate(values, start=1):
        if value >= cutoff:
            result.append((position,value))
    return result
````

````python
# Question 9 public tests
assert passing_positions([65, 90, 45, 82, 73], 70) == [(2, 90), (4, 82), (5, 73)]
assert passing_positions([], 70) == []
print("Question 9 tests passed.")
````

## Question 10 — `while` loop and variable update (6 points)

Write `countdown_stock(start, step)`.

- Begin with `current = start`.
- While `current > 0`, append `current` to a list.
- Subtract `step` from `current` on every iteration.
- Return the list.
- Assume `step` is positive.

Example:

```python
countdown_stock(11, 3) == [11, 8, 5, 2]
```


````python
# Question 10 code
def countdown_stock(start, step): # [11, 3]
    current = start
    result = []
    while current > 0 :
        result.append (current)
        current = current - step
    return result
print (countdown_stock)
````

````python
# Question 10 public tests
assert countdown_stock(11, 3) == [11, 8, 5, 2]
assert countdown_stock(0, 3) == []
assert countdown_stock(1, 5) == [1]
print("Question 10 tests passed.")
````

## Question 11 — `break`, `continue`, filtering, and average (8 points)

    Write `process_tickets(values)`.

    Process from left to right:

    - When `999` appears, use `break`; ignore all later values.
    - For a negative value, use `continue`.
    - Accept values greater than or equal to `50`.
    - Build `accepted`, calculate `total` with a running total, and calculate `average`.
    - If no values are accepted, average is `0`.
    - Return `(accepted, total, average)`.

    Trace this input before coding:

    ```python
    [55, -3, 42, 80, 999, 100]
    ```

    | value | action | accepted after turn | total after turn |
    |---:|---|---|---:|
    | 55 |  |  |  |
| -3 |  |  |  |
| 42 |  |  |  |
| 80 |  |  |  |
| 999 |  |  |  |


````python
# Question 11 code
def process_tickets(values):
    accepted =[]
    total = 0
    average = 0
    for value in values:
        if value == 999:
           break 
        elif value <0:
            continue
        elif value >= 50:
            accepted.append(value)
            total = sum(accepted)
            average = total / len(accepted)
    return (accepted, total, average)
````

````python
# Question 11 public tests
assert process_tickets([55, -3, 42, 80, 999, 100]) == ([55, 80], 135, 67.5)
assert process_tickets([999]) == ([], 0, 0)
assert process_tickets([-2, 50, 999, 60]) == ([50], 50, 50.0)
print("Question 11 tests passed.")
````

## Question 12 — Iterating over dictionaries (8 points)

Write `purchase_total(prices, quantities)`.

- Loop through `quantities.items()`.
- Ignore an item that is missing from `prices`.
- Ignore a quantity that is `0` or negative.
- For each valid item, calculate `prices[item] * quantity`.
- Store each item cost in a dictionary named `details`.
- Add each item cost to a running `total`.
- Return `(details, total)`.

Starter data:

```python
prices = {'tea': 3, 'wrap': 7, 'fruit': 2}
quantities = {'tea': 2, 'wrap': 1, 'cookie': 4, 'fruit': 0}
```

Before coding, predict `details` and `total`.


````python
# Question 12 code
def purchase_total(prices, quantities):
    details = {}
    total= 0
    for item, quantity in quantities.items():
        if item not in prices:
            continue
        if quantity <= 0:
            continue
        cost = prices[item] * quantity
        details[item] = cost
        total +=  cost
    return (details,total)
````

````python
# Question 12 public tests
prices = {'tea': 3, 'wrap': 7, 'fruit': 2}
quantities = {'tea': 2, 'wrap': 1, 'cookie': 4, 'fruit': 0}
assert purchase_total(prices, quantities) == ({'tea': 6, 'wrap': 7}, 13)
assert purchase_total(prices, {}) == ({}, 0)
print("Question 12 tests passed.")
````

## Question 13 — Nested loops (8 points)

Write `multiplication_grid(rows, columns)`.

- The outer loop counts rows from `1` through `rows`.
- The inner loop counts columns from `1` through `columns`.
- Append `(row, column, row * column)` for every pair.
- Return the completed list.

Required example:

```python
multiplication_grid(2, 3)
```

should return:

```python
[(1, 1, 1), (1, 2, 2), (1, 3, 3), (2, 1, 2), (2, 2, 4), (2, 3, 6)]
```


````python
# Question 13 code
def multiplication_grid(rows, columns):
    grid = []
    for row in range(1, rows+ 1):
        for column in range(1,columns +1):
            grid.append((row,column,row * column))
    return grid
````

````python
# Question 13 public tests
assert multiplication_grid(2, 3) == [(1, 1, 1), (1, 2, 2), (1, 3, 3), (2, 1, 2), (2, 2, 4), (2, 3, 6)]
assert multiplication_grid(0, 3) == []
assert multiplication_grid(1, 1) == [(1, 1, 1)]
print("Question 13 tests passed.")
````

## Question 14 — pandas DataFrame row iteration (8 points)

The lecture shows `df.iterrows()` for visiting rows.

Write `label_visits(df, cutoff)`.

- Create an empty list named `labels`.
- Loop with `for index, row in df.iterrows():`.
- If `row['visits'] >= cutoff`, append `'Frequent'`; otherwise append `'Review'`.
- Assign the list to `df["result"]`.
- Return `df`.

Starter data:

```python
{'name': ['Ava', 'Noah', 'Mia'], 'visits': [8, 3, 6]}
```

Predict the new `result` column before coding: __________________________


````python
# Question 14 code
import pandas as pd

def label_visits(df, cutoff):
    labels =[]
    for index, row in df.iterrows():
        if row['visits'] >= cutoff:
           labels .append('Frequent')
        else:        
            labels.append('Review')
    df["result"] = labels
    return df      
````

````python
# Question 14 public tests
df_test = pd.DataFrame({'name': ['Ava', 'Noah', 'Mia'], 'visits': [8, 3, 6]})
df_result = label_visits(df_test, 6)
assert list(df_result["result"]) == ['Frequent', 'Review', 'Frequent']
assert list(df_result.columns) == ["name", "visits", "result"]
print("Question 14 tests passed.")
````
