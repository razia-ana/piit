## Lesson 2 — Understanding `range()`

`range()` produces a sequence of whole numbers. The stopping number is **not**
included.

- `range(stop)` starts at `0`.
- `range(start, stop)` starts where you tell it.
- `range(start, stop, step)` also controls the amount of each jump.


````python
# range(stop)
for number in range(5):
    print(number)
````

````python
# range(start, stop)
for number in range(1, 6):
    print(number)
````

````python
# range(start, stop, step)
for number in range(2, 11, 2):
    print(number)
````

````python
# A negative step counts backward.
for number in range(5, 0, -1):
    print(number)

print("Go!")
````

## Lesson 3 — Calculations and decisions

You can perform calculations inside a loop. `** 2` means “raised to the power
of two,” or squared.


````python
for number in range(1, 6): #[1,2,3,4,5]
    square = number ** 2
    print(number, "squared is", square)
````

````python
number = 8

for multiplier in range(1, 11): # [1......10]
    answer = number * multiplier
    print(number, "x", multiplier, "=", answer)
````

### A running total

Start `total` at zero. On every turn, add the current number to it.

`total += number` is a shorter form of `total = total + number`.


````python
total = 0

for number in range(1, 6): # [1,2,3,4,5]
    total = total + number
    print("After adding", number, "the total is", total)

print("Final total:", total)
````

### Using `if` inside a loop

The `%` operator finds the remainder. An even number has a remainder of zero
when divided by two.


````python
for number in range(1, 21):
    if number % 2 == 0:           
        print(number, "is even")
    else:
        print(number, "is odd")
````

## Lesson 4 — Looping through lists

A list stores several values in order. Square brackets create a list.

On each turn, `fruit` temporarily stores the next item.


````python
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print(fruit)

````

### Building a new list with `.append()`


````python
squares = []

for number in range(1, 6): #[1,2,3,4,5]
    squares.append(number ** 2) # ** to the power

print(squares)
````

### Getting positions with `enumerate()`


````python
fruits = ["apple", "banana", "mango"]

for position, fruit in enumerate(fruits, start=1):
    print(position, fruit)
````

## Lesson 5 — `while` loops

A `while` loop repeats while its condition remains `True`. Use it when you do
not know in advance how many repetitions will be needed.

> **Warning:** If the condition never becomes `False`, the loop runs forever.
> In Jupyter, press the stop button or choose **Kernel → Interrupt Kernel**.


````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

````python
count = 6

while count <= 5:
    print(count)
    count += 1
````

### Countdown example


````python
count = 5

while count >= 1:
    print(count)
    count -= 1

print("Go!")
````

````python
count = 6

while count >= 3:
    print(count)
    count -= 1

print("Go!")
````

## Lesson 6 — `break` and `continue`

- `break` stops a loop immediately.
- `continue` skips the rest of the current turn.


# # querry for rasul bhai

````python
# break example without user input
commands = ["hello", "again", "quit", "ignored"]

for command in commands:
    if command == "quit":
        break
    print("Processing:", command)

print("Loop stopped")
````

````python
commands = ["hello", "again", "quit", "ignored"]

for command in commands:
    if command == "quit":
        continue
    print("Processing:", command)

print("Loop stopped")
````

````python
# continue skips the number 3
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
````

## Lesson 7 — Dictionaries

A dictionary stores **key-value pairs**. In this example, each student name is
a key and each score is its value.

`.items()` gives both pieces to the loop.


````python
scores = {
    "Ana": 90,
    "Ben": 65,
    "Chris": 82
}
for name, score in scores.items():
    print(name, score)
````

````python
for name, score in scores.items():
    if score >= 70:
        print(name, "passed")
    else:
        print(name, "needs more practice")
````

## Lesson 8 — Nested loops

A nested loop is a loop inside another loop. The inner loop completes all its
turns for every turn of the outer loop.


````python
for row in range(1, 5):
    for column in range(1, 3):
        print("Row:", row, "Column:", column)
````

````python
# A small multiplication table
for first in range(1, 4):
    for second in range(1, 4):
        print(first, "x", second, "=", first * second)
    print()
````

## Lesson 9 — pandas DataFrames

A pandas DataFrame is a table with rows and columns. Run the import cell first.


````python
import pandas as pd

df = pd.DataFrame({
    "name": ["Ana", "Ben", "Chris"],
    "score": [90, 65, 82]
})
df
````

### Looping through DataFrame rows


````python
for index, row in df.iterrows():
    print(row["name"], row["score"])
````

### The pandas way

For ordinary data analysis, column operations are normally clearer and faster
than row-by-row loops.


````python
df["passed"] = df["score"] >= 70
passing_students = df[df["score"] >= 70]

passing_students
````

````python
df["result"] = df["score"].apply(
    lambda score: "Pass" if score >= 70 else "Review"
)

df
````

## Common mistakes

| Symptom | Likely cause | Fix |
|---|---|---|
| `IndentationError` | The loop body is not indented | Use four spaces inside the loop |
| `SyntaxError` near the loop | The colon is missing | Add `:` after the `for` or `while` statement |
| Last number is missing | `range()` excludes its stop value | Use one more than the final number wanted |
| Infinite `while` loop | The condition never becomes `False` | Update the control variable or use `break` |
| `NameError` | A variable was not created or its cell was not run | Run the defining cell or restart and run all |


## Final project — Student Grade Manager

Complete the project in stages:

1. Create a dictionary for student scores.
2. Loop through it and label every student `Pass` or `Review`.
3. Calculate the class total and average.
4. Find the student with the highest score.
5. Convert the results to a DataFrame.


````python
# Starter data; replace or add students.
students = {
    "Ava": 84,
    "Noah": 67,
    "Mia": 93,
    "Leo": 72
}

total = 0
passing_students = []

for name, score in students.items():
    total += score

    if score >= 70:
        status = "Pass"
        passing_students.append(name)
    else:
        status = "Review"

    print(name, "-", score, "-", status)

average = total / len(students)
highest_student = max(students, key=students.get)

print("\nClass average:", average)
print("Highest-scoring student:", highest_student)
print("Passing students:", passing_students)
````

````python
# Convert the project results into a DataFrame.
results_df = pd.DataFrame(
    students.items(),
    columns=["name", "score"]
)

results_df["result"] = results_df["score"].apply(
    lambda score: "Pass" if score >= 70 else "Review"
)

results_df
````

## Loop decision guide

| What you need | Tool to consider |
|---|---|
| Repeat a known number of times | `for` with `range()` |
| Visit every item in a list | `for` |
| Continue until a condition changes | `while` |
| Stop early | `break` |
| Skip one turn | `continue` |
| Visit dictionary keys and values | `for ... in dictionary.items()` |
| Visit DataFrame rows | `df.iterrows()`—only when truly needed |
