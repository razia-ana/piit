```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
 # 1. Histogram
df['mpg'].plot(kind ='hist', title='Distribution of mpg')
plt.show()
```

```python
# 2. Scatter Plot
df.plot(kind='scatter', x='hp', y='mpg', title='HP vs MPG')
plt.show()
```

```python
#3 Boxplot
df.boxplot(column='mpg',by = 'cyl')
plt.suptitle('')
plt.show()
```

```python
#4 line plot
df.sort_values('wt').plot(x='wt', y='mpg', kind='line', title ='Weight vs MPG')
plt.show()
```

```python
df.plot(kind='bar')
```

```python
df.plot(kind= 'density')
```

```python
# 1. Histogram
plt.figure()
plt.hist(df['mpg'])
plt.title('Distribution of Mpg')
plt.xlabel('mpg')
plt.ylabel('distribution')
plt.show() # MPG distribution: Mostly clustered between ~18–24 → typical mid-efficiency cars.
```

```python
#2 scatter
plt.figure()
plt.title('HP vs MPG')
plt.scatter(df['hp'],df['mpg'])
plt.xlabel('Horsepower')
plt.ylabel('mpg')
plt.show()
```

```python
#3 boxplot
plt.figure()
df.boxplot(column ='mpg', by='cyl')
plt.title('MPG by Cylinders')
plt.suptitle('')
plt.show()
```

```python
#line plot
df_sorted=df. sort_values(by='wt')
plt.figure()
plt.plot(df_sorted['wt'], df_sorted['mpg'])
plt.title('Weight vs Mpg')
plt.xlabel('Weight')
plt.ylabel('mpg')
```

```python
import pandas as pd
import numpy as np
from pydataset import data
```

```python
df=data('iris')
df.head()
```

```python
plt.hist(df['Petal.Length'],bins=20, color='purple',edgecolor='white')
plt.title('Distribution of Petal Lengths')
plt.xlabel('Length(cm)')
plt.ylabel('Number of flowers')
plt.grid(axis='y', alpha= 0.3)
plt.show()
```

```python
plt.hist(df['Petal.Length'],bins=10, color= 'red',edgecolor= 'green')
plt.title('Distribution of Petal Length')
plt.xlabel('Petal Length')
plt.ylabel ('How many Flowers')
plt.show()
```

```python
plt.scatter(df['Petal.Length'],df['Petal.Width'], c='forestgreen', s=40)
plt.title('Petal Length vs Width')
plt.xlabel('length(cm)')
plt.ylabel('Width')
plt.show()
```
