## Indexing and slicing

Python uses zero-based indexing. Negative indexes count backward from the end.

````python
text = "Bootcamp"
text1 = 'Friday'
print(text1[0])
print(text[0])      # first character B
print(text[-1])     # last character p
print(text[0:4])    # characters 0 through 3 Bootc/Bo
print(text[4:])     # character 4 to the end tcamp/camp
print(text[::-1])   # reversed copy pmactooB
````

# 2. Formatting text with values

Formatting lets us build readable messages from variables. The source notebooks introduce `%` specifiers and `.format()`. This tutorial also includes **f-strings**, the preferred style in modern Python.

## 2.1 Percent-style formatting

````python
category = "Z"
quantity = 500
price = 60.359000

print('This item is from category %s.' % category)
print("We currently have %d units." % quantity)
print("This stock costs $%.2f per share." % price)
print('Category "%s" has %d units at $%.2f each.' % (category, quantity, price))
````

Common specifiers:

| Specifier | Meaning | Example |
|---|---|---|
| `%s` | string representation | `%s` |
| `%d` | integer | `%d` |
| `%f` | floating-point number | `%f` |
| `%.2f` | float with two decimal places | `60.35` |

## 2.2 The `.format()` method

````python
company = "Company B"
quarter = 3
prediction = "increase"

message = "{} expects sales in Q{} to {}.".format(company, quarter, prediction)
print(message)

named_message = "{company} expects sales in Q{quarter} to {prediction}.".format(
    company=company,
    quarter=quarter,
    prediction=prediction,
)
print(named_message)
````

## 2.3 F-strings

````python
shop = "Fast Cars"
brand = "Audi"
units = 1000000
model_price = 37500.50

print(f'The "{shop}" shop expects to sell {units:,} units of the new {brand} model.')
print(f"Estimated model price: ${model_price:,.2f}")
````

Useful f-string formatting:

- `{value:.2f}` → two decimal places;
- `{value:,}` → thousands separator;
- `{value:,.2f}` → both thousands separator and two decimals;
- `{value:.1%}` → percentage formatting for a decimal value.

````python
revenue = 1250000.5
growth_rate = 0.0734

print(f"Revenue: ${revenue:,.2f}")
print(f"Growth: {growth_rate:.1%}")
````

# 4. Core string methods

String methods are functions attached to string objects. Most return a new value.

## 4.1 `replace()`, `startswith()`, and `endswith()`

````python
s = "Administrative expenses"
s1 = s.replace("Administrative", "Selling")

print(s1)
print(s.startswith("Admin"))
print(s1.startswith("Sell"))
print(s1.startswith("sell"))   # case-sensitive
print(s1.endswith("ses"))
````

## 4.2 Changing letter case

````python
name = "mRs. aMY mOORE"

print(name.lower())
print(name.upper())
print(name.capitalize())
print(name.title())
print(name.swapcase())
````

**Caution:** `.title()` is convenient but not perfect for every real-world name or brand. Always inspect results when capitalization matters.

## 4.3 Removing surrounding characters

````python
report = "   quarterly earnings report   "

print(repr(report.strip()))
print(repr(report.lstrip()))
print(repr(report.rstrip()))
````

The argument to `strip(chars)` is a **set of characters**, not an exact prefix or suffix.

````python
value = "www.example.com"
print(value.strip("w.com"))

# For exact prefixes/suffixes, prefer these methods:
print(value.removeprefix("www."))
print(value.removesuffix(".com"))
````

## 4.4 Finding and counting text

````python
sentence = "Data cleaning turns raw data into useful data."

print(sentence.find("data"))       # first exact lowercase match
print(sentence.lower().find("data"))
print(sentence.lower().count("data"))
print("raw" in sentence)
print("missing" not in sentence)
````

# 5. Splitting, joining, and parsing text

## 5.1 `split()`

````python
s2 = "personal ID, first name, last name, date of birth, place of birth, city of birth"

fields = s2.split(",")
print(fields)
print(len(fields))

clean_fields = [field.strip() for field in fields]
print(clean_fields)
````

With no argument, `.split()` separates on runs of whitespace. With a delimiter, it separates wherever that exact delimiter appears.

````python
phrase = "Price per unit"
print(phrase.split())
print(phrase.split("per"))
print(phrase.split(" ", maxsplit=1))
````

## 5.2 `join()`

````python
words = ["quarterly", "earnings", "report"]
print(" ".join(words))
print("-".join(words))
````

# 6. Reusable cleaning functions

When the same cleaning steps appear repeatedly, place them in a function. Small functions are easier to test and reuse.

````python
def normalize_label(value: str) -> str:
    """Remove outside whitespace, collapse internal whitespace, and title-case text."""
    words = value.strip().split()
    return " ".join(words).title()

examples = [
    "  customer   success ",
    "DATA engineering",
    " product   management  ",
]

for item in examples:
    print(normalize_label(item))
````

## Defensive cleaning

Real datasets often contain missing values or non-string objects. A robust helper should define what happens in those cases.

````python
def safe_normalize_label(value):
    if pd.isna(value):
        return pd.NA
    return normalize_label(str(value))

for item in ["  data science ", None, 101]:
    print(repr(safe_normalize_label(item)))
````

# 7. Text data in pandas

The pandas `.str` accessor applies string operations element by element to a Series. These operations are usually clearer and faster than manually looping through rows.

````python
operational_kpis = pd.Series(
    ["employee satisfaction rating", "employee churn rate"],
    name="kpi",
)
operational_kpis
````

## 7.1 Applying string methods with `.str`

````python
print(operational_kpis.str.upper())
print()
print(operational_kpis.str.replace("employee ", "", regex=False))
print()
print(operational_kpis.str.startswith("employee"))
````

## 7.2 String indexing and slicing

````python
print(operational_kpis.str[0])
print()
print(operational_kpis.str[:8])
````

## 7.3 Splitting a Series

````python
names = pd.Series(["Martin Peterson", "John Fitzpatrick", "Kate Cruz"])

parts = names.str.split(" ", n=1, expand=True)
parts.columns = ["first_name", "last_name"]
parts
````

## 7.4 Searching safely

````python
last_names = parts["last_name"]

print(last_names.str.contains("s", case=False, regex=False))
print(last_names.str.len())
````

**Regex note:** `.str.contains()` treats its pattern as a regular expression by default. Use `regex=False` when searching for literal punctuation such as `$`, `+`, `?`, or `.`.

````python
house_prices = pd.Series(["$400,000", "$500,000", "600,000"])
print(house_prices.str.contains("$", regex=False))
````

## 7.5 Missing and mixed values

````python
mixed = pd.Series([" Text_data ", 34, None], dtype="object")

print(mixed)
print()
print(mixed.str.strip())
````

# 8. Guided lab: Clean messy customer data

We will clean a small customer table containing inconsistent names, emails, cities, phone numbers, and status labels.

````python
customers = pd.DataFrame({
    "customer_id": [" C-001 ", "c-002", " C-003", "c-004 "],
    "name": ["  ada lovelace", "GRACE   HOPPER ", "alan Turing", " Katherine johnson  "],
    "email": [" ADA@EXAMPLE.COM ", "grace@example.com", " alan@EXAMPLE.COM", None],
    "city": [" london", "NEW YORK ", "Manchester", "  hampton"],
    "status": [" active ", "ACTIVE", "inactive ", "Active"],
    "phone": ["+44 20 0000 0001", "+1 (212) 555-0199", "+44-161-555-0102", "(757) 555 0110"],
})

customers
````

## Step 1: inspect before changing

````python
print(customers.dtypes)
print()
print(customers.isna().sum())
````

## Step 2: clean identifiers and labels

````python
clean_customers = customers.copy()

clean_customers["customer_id"] = clean_customers["customer_id"].str.strip().str.upper()
clean_customers["name"] = (
    clean_customers["name"]
    .str.strip()
    .str.split()
    .str.join(" ")
    .str.title()
)
clean_customers["city"] = clean_customers["city"].str.strip().str.title()
clean_customers["status"] = clean_customers["status"].str.strip().str.lower()
clean_customers
````

## Step 3: normalize email and phone fields

````python
clean_customers["email"] = clean_customers["email"].str.strip().str.lower()

clean_customers["phone_digits"] = clean_customers["phone"].str.replace(
    r"\D+", "", regex=True
)

clean_customers[["email", "phone", "phone_digits"]]
````

## Step 4: derive useful columns

````python
clean_customers["email_domain"] = clean_customers["email"].str.split("@").str[-1]
clean_customers["is_active"] = clean_customers["status"].eq("active")

clean_customers
````

## Step 5: validate assumptions

````python
assert clean_customers["customer_id"].str.match(r"C-\d{3}").all()
assert set(clean_customers["status"]) <= {"active", "inactive"}
assert clean_customers["name"].str.contains(r"\s{2,}", regex=True).sum() == 0

print("Validation checks passed.")
````

# 12. Text-method cheat sheet

| Goal | Python string | pandas Series |
|---|---|---|
| remove outside whitespace | `s.strip()` | `series.str.strip()` |
| lowercase | `s.lower()` | `series.str.lower()` |
| title case | `s.title()` | `series.str.title()` |
| replace literal text | `s.replace(a, b)` | `series.str.replace(a, b, regex=False)` |
| split text | `s.split(delimiter)` | `series.str.split(delimiter)` |
| join words | `delimiter.join(words)` | `series.str.join(delimiter)` |
| check prefix | `s.startswith(prefix)` | `series.str.startswith(prefix)` |
| check suffix | `s.endswith(suffix)` | `series.str.endswith(suffix)` |
| search literal text | `text in s` | `series.str.contains(text, regex=False)` |
| character count | `len(s)` | `series.str.len()` |
| access character/slice | `s[0]`, `s[:3]` | `series.str[0]`, `series.str[:3]` |
